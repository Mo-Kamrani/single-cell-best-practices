# scRNA-seq Feature Selection

This notebook continues from the normalization workflow and identifies informative genes using **binomial deviance**.

Place the normalized input dataset at:

```text
data/s4d8_normalization.h5ad
```

The workflow:

1. Loads the normalized `AnnData` object.
2. Transfers the unnormalized count matrix from `adata.X` to R.
3. Computes binomial deviance with `scry`.
4. Marks the 4,000 most deviant genes.
5. Visualizes the selected genes using mean and dispersion.
6. Saves the updated object to `results/s4d8_feature_selection.h5ad`.

The Python environment requires `scanpy`, `rpy2`, `numpy`, `matplotlib`, and `seaborn`.  
The R environment requires `scry`, `SingleCellExperiment`, and `Matrix`.


## Environment setup

````python
# Import Python and R interoperability packages
import logging
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import rpy2.rinterface_lib.callbacks as rcb
import rpy2.robjects as ro
import rpy2.robjects.packages as rpackages
import scanpy as sc
import seaborn as sns
from rpy2.robjects import default_converter, numpy2ri, pandas2ri, r
from rpy2.robjects.conversion import localconverter

# Suppress verbose Scanpy logging
sc.settings.verbosity = 0

# Use clean, minimal plotting defaults
sc.settings.set_figure_params(
    dpi=80,
    facecolor="white",
    frameon=False,
)

# Suppress verbose rpy2 callback logging
rcb.logger.setLevel(logging.ERROR)
````

````python
# Enable R cells in Jupyter
%load_ext rpy2.ipython
````

````python
%%R
# Load the R packages used for deviance-based feature selection
library(scry)
library(SingleCellExperiment)
````

## Load the normalized dataset

````python
# Load the AnnData object created in the normalization notebook
DATA_PATH = Path("data/s4d8_normalization.h5ad")

adata = sc.read_h5ad(DATA_PATH)
adata
````

`adata.X` still contains the unnormalized SoupX-corrected counts.  
The normalized matrices remain available in separate layers.


## Transfer the count matrix to R

````python
# Transpose the count matrix so that genes are rows and cells are columns
X_sparse = adata.X.T.tocoo()

# Import the R Matrix package
Matrix = rpackages.importr("Matrix")

# Transfer cell and gene metadata to R
with localconverter(
    ro.default_converter
    + pandas2ri.converter
    + numpy2ri.converter
):
    ro.globalenv["obs"] = adata.obs
    ro.globalenv["var"] = adata.var

# Extract nonzero coordinates and values from the COO matrix
i, j = X_sparse.row, X_sparse.col
x = X_sparse.data

# R uses one-based indexing
ro.globalenv["i"] = ro.IntVector((i + 1).tolist())
ro.globalenv["j"] = ro.IntVector((j + 1).tolist())
ro.globalenv["x"] = ro.FloatVector(x.tolist())

# Reconstruct the sparse count matrix in R
r(
    "X <- sparseMatrix(i = i, j = j, x = x, dims = c({}, {}))".format(
        *X_sparse.shape
    )
)
````

## Calculate binomial deviance

````python
%%R
# Build a SingleCellExperiment object and calculate deviance for each gene
sce <- SingleCellExperiment(
  assays = list(X = X),
  colData = obs,
  rowData = var
)

sce <- devianceFeatureSelection(
  sce,
  assay = "X"
)
````

````python
# Transfer the binomial deviance values from R back to Python
with localconverter(
    default_converter
    + pandas2ri.converter
    + numpy2ri.converter
):
    binomial_deviance = ro.r(
        "rowData(sce)$binomial_deviance"
    )
````

## Select the 4,000 most deviant genes

````python
# Identify the indices of the 4,000 genes with the largest deviance
idx = binomial_deviance.argsort()[-4000:]

# Create a Boolean mask for the selected genes
mask = np.zeros(
    adata.var_names.shape,
    dtype=bool,
)
mask[idx] = True

# Store both the selection mask and the deviance values
adata.var["highly_deviant"] = mask
adata.var["binomial_deviance"] = binomial_deviance
````

## Visualize the selected features

````python
# Calculate mean expression and dispersion using Scran-normalized values
sc.pp.highly_variable_genes(
    adata,
    layer="scran_normalization",
)
````

````python
# Plot gene dispersion against mean expression
ax = sns.scatterplot(
    data=adata.var,
    x="means",
    y="dispersions",
    hue="highly_deviant",
    s=5,
)

ax.set_xlim(None, 1.5)
ax.set_ylim(None, 3)

plt.show()
````

## Save the feature-selected dataset

````python
# Save the AnnData object with feature-selection annotations
OUTPUT_PATH = Path("results/s4d8_feature_selection.h5ad")
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

adata.write_h5ad(OUTPUT_PATH)
````
